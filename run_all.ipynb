## Settings

Keep the defaults to rebuild the complete research workflow from saved raw data.
`REUSE_FORECASTS=True` skips build notebooks (including SHAP regeneration), so those sources
will be marked as not executed in the final index. Keep `USE_FIT_CACHE=True` when reusing forecasts.
Set `USE_FIT_CACHE=False` for fresh GARCH fits when rebuilding forecasts.

`START="fetch"` and `INCLUDE_DATA_CHECKS=True` independently require internet and a
`FRED_API_KEY` in the environment or repository `.env`. The optional checks compare market
series with ALFRED first releases and examine BAA10Y vintages. They run after macro transformation.


```python
from pathlib import Path
import sys

ROOT = next(
    (p for p in (Path.cwd(), *Path.cwd().parents)
     if (p / "run_pipeline.py").is_file() and (p / "models/research_config.py").is_file()),
    None,
)
if ROOT is None:
    raise FileNotFoundError("Start this notebook's kernel from inside this project checkout.")
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

import run_pipeline as pipeline

START = "raw"                  # "fetch" downloads fresh macro data (requires FRED_API_KEY).
REUSE_FORECASTS = False         # True skips b0-b3 and reads saved main forecasts and Elastic Net rolling forecasts.
USE_FIT_CACHE = True            # GARCH fits use the existing input-sensitive cache.
INCLUDE_DATA_CHECKS = False     # True runs both online FRED/ALFRED diagnostic notebooks.
RUN_DIR = None                 # Or a NEW directory; existing run directories are never reused.

DISPLAY_MODE = "compact"
```

## Preview the execution order

Data preparation → EDA → b0–b3 → e1–e6 → all forecast robustness → p1–p3 → rp1–rp2.
The two optional online data checks are inserted after macro transformation when enabled.
This cell lists steps without executing them.


```python
steps = pipeline.pipeline_steps(
    reuse_forecasts=REUSE_FORECASTS,
    include_data_checks=INCLUDE_DATA_CHECKS,
)
print(f"Python: {sys.executable}")
print(f"Mode: {DISPLAY_MODE}; {len(steps)} sequential steps; macro start: {START}")
for i, filename in enumerate(steps, 1):
    labels = pipeline.report_labels(filename)
    suffix = " — " + "; ".join(labels) if labels else ""
    print(f"{i:02d}. {filename}{suffix}")
```

## Execute the workflow

Stages run sequentially in separate kernels using the Python interpreter shown above.
Forecast checks run before evaluation. The runner stops at the first failure and saves
the failed notebook or script log. The final index links to executed copies and available data.


```python
run_dir = pipeline.run_pipeline(
    start=START,
    reuse_forecasts=REUSE_FORECASTS,
    use_fit_cache=USE_FIT_CACHE,
    include_data_checks=INCLUDE_DATA_CHECKS,
    run_dir=RUN_DIR,
    display_mode=DISPLAY_MODE,
)
```
